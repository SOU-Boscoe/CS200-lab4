## A note about `main()`

In Chapter 5 of the zyBook, you will see programs written with a `main()` function that is called at the bottom of the file, like this:

```python
def main() :
   result = cubeVolume(2)
   print(result)

def cubeVolume(sideLength) :
   return sideLength ** 3

main()
```

In a Jupyter notebook, **we don't need `main()`**. We define a function in one cell, run that cell, and then call the function from a cell below it. Just remember: **a function must be defined (its cell must be run) before you can call it.** Otherwise you get a `NameError`, the same "run the cells in order" issue from Lab 2's Jupyter quiz question.

````python
def boxString(contents) :
   n = len(contents)
   print("-" * (n + 2))
   print("!" + contents + "!")
   print("-" * (n + 2))

boxString("Hello")
boxString("CS 200")
````
